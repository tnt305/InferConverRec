```python
import re

predictions = [
    "system [movie] is classic",
    "system [movie] is about [guy] who is trying to get [job] in [hotel]",
    "system [movie] is good [action] [film]",
    "system hi i'm looking for good [action] [movie]",
    "system [movie] is [comedy]"
]

def compute_item_ratio(strs):
    item_ratio = 0
    for str in strs:
        items = re.findall(r'[movie]', str)
        item_ratio
        item_ratio += len(items)
    return item_ratio

result = compute_item_ratio(predictions)
print(f"Total number of items: {result}")
print(f"Average items per prediction: {result / len(predictions):.2f}")
```

```python
text = 'system <movie> was good one'


import re


re_art = re.compile(r'\b(a|an|the)\b')
re_punc = re.compile(r'[!"#$%&()*+,-./:;=?@\[\]\\^`{|}~_\']')
re_space = re.compile(r'\s+')

def normalize_answer(s):
    """
    Lower text and remove punctuation, articles and extra whitespace.
    """
    s = s.lower()
    s = re_punc.sub(' ', s)
    s = re_art.sub(' ', s)
    s = re_space.sub(' ', s)
    s = ' '.join(s.split())
    return s


normalize_answer(text)
```
